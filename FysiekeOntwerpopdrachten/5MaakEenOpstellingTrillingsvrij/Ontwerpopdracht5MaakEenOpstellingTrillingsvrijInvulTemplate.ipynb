## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
omega = np.linspace(0, 100,int(1e6)) #omega array
omega_0= np.sqrt((10*np.pi)**2/4) #cut of angular freq
print('the cuttoff frequency is',int(omega_0*1000)/1000,'rad/s') 
A_ratio= 1/(1-(omega**2)/(omega_0**2)) #amplitude ratio
```

```python
# Voeg hier je plot in.
plt.figure()
plt.plot(np.log(omega),np.log(np.abs(A_ratio)),'r--') #ploting on a log-log scale
plt.xlabel('Log(Angular Frequency (Rad*s^-1))')
plt.ylabel('Log(Amplitude Ratio)')
plt.title('Angular Frequency vs the Amplitude Ratio on a Log-Log Plot')
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g=9.81
d_x=g/(omega_0**2)#delta x min
print(f'delta x minimum {d_x:.4f}m')
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
m = np.linspace(0.1,2,20)
c = (m*g)/d_x
"""c2 = (m*g)/(2*d_x)
c5 = (m*g)/(5*d_x)"""
plt.figure()
plt.plot(m,c,'r--', label='1*d_x') #normal d_x
plt.plot(m,c*2,'b--', label='2*d_x') #2x d_x
plt.plot(m,c*5,'k--', label='5*d_x') #5x d_x
plt.xlabel('Mass of Object (kg)')
plt.ylabel('Spring Constant (N/m)')
plt.legend()
plt.title('Mass vs Spring Constant')
plt.savefig('massa.png')
plt.show()
```
